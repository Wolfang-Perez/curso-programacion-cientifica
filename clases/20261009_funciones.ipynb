```python
def maximo_valor(datos2): #Es necesario def y un nombre mas los parametros de entrada que puede estar vacío y ":"
  maximo_valor = -1000
  for dato in datos2:
    if dato >= maximo_valor:
      maximo_valor = dato
  print(f"El máximo valor es {maximo_valor}")
  #Un return que sería el retorno y es opcional.
```

```python
import random

datos = [random.randint(0,10000) for _ in range(10)]
print(datos)
maximo_valor(datos)
```

```python
def g(x = 1,y = 2):
  return x**2 + y**2
```

```python
resultado_1 = g(2,3)
resultado_2 = g(2,5)
resultado_3 = g()

print(f"Resultado 1: {resultado_1}")
print(f"Resultado 2: {resultado_2}")
print(f"Resultado 3: {resultado_3}")

```
