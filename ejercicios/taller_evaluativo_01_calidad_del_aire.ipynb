## Ejercicio 2 — Conversión de unidades y error

El sensor mide en mg/m3 y hay que pasarlo a ug/m3 multiplicando por 1000
(el `factor_conversion` del Ejercicio 1).

Con la lectura ya convertida se calculan tres errores frente al valor del
patrón: error absoluto, error relativo en porcentaje y error al cuadrado.

```python
lectura_sensor_mg = 0.0418
valor_patron_ug = 40.0

lectura_sensor_ug = lectura_sensor_mg * factor_conversion

error_absoluto = lectura_sensor_ug - valor_patron_ug
error_relativo_porcentual = (error_absoluto / valor_patron_ug) * 100
error_cuadratico = error_absoluto ** 2

print("Lectura del sensor (ug/m3):", lectura_sensor_ug)
print("Error absoluto (ug/m3):", error_absoluto)
print("Error relativo porcentual (%):", error_relativo_porcentual)
print("Error cuadratico:", error_cuadratico)
```

## Ejercicio 3 — Cobertura temporal

La estación toma 12 lecturas por hora. Con las 305 lecturas se calcula

## Ejercicio 4 — Validar una lectura

Se revisa si la lectura 41.8 sirve o no. Una lectura sirve si no es el
centinela -999.0 y si está dentro del rango de 0 a 500.

Se calculan tres booleanos y se muestran con `print()`. Aquí no se usa `if`.

```python
lectura = 41.8
codigo_dato_faltante = -999.0
limite_superior_sensor = 500.0

es_dato_faltante = lectura == codigo_dato_faltante

esta_en_rango = lectura >= 0 and lectura <= limite_superior_sensor

lectura_valida = esta_en_rango and not es_dato_faltante

print("Es dato faltante:", es_dato_faltante)
print("Esta en rango:", esta_en_rango)
print("Lectura valida:", lectura_valida)
```

## Ejercicio 6 — Limpiar la serie horaria

La lista tiene 12 lecturas, pero dos son -999.0 (dato faltante) y hay que
sacarlas.

Se recorre la lista con un `for`. Cuando aparece un -999.0 se cuenta como
descartado y se usa `continue` para saltarlo. Las demás se guardan en
`lecturas_validas` y se van sumando.

Al final se calcula el promedio y se muestra cuántas válidas quedaron,
cuántas se descartaron y el promedio.

```python
lecturas_pm25 = [8.4, 12.7, -999.0, 15.2, 41.8, 9.6, -999.0, 58.3, 22.0, 13.9, 7.5, 36.1]

lecturas_validas = []
suma_pm25 = 0.0
lecturas_descartadas = 0

for lectura_actual in lecturas_pm25:
    if lectura_actual == -999.0:
        lecturas_descartadas += 1
        continue
    lecturas_validas.append(lectura_actual)
    suma_pm25 += lectura_actual

promedio_pm25 = suma_pm25 / len(lecturas_validas)

print("Lecturas validas:", len(lecturas_validas))
print("Lecturas descartadas:", lecturas_descartadas)
print("Promedio PM2.5:", promedio_pm25)
```

## Ejercicio 7 — Estadística de la serie limpia

Con las lecturas válidas y el promedio del Ejercicio 6 se calcula el máximo,
el mínimo, el rango y la desviación estándar.

Se usan dos `for`: uno para el máximo y el mínimo, y otro para la desviación
(que necesita el promedio ya calculado).

```python
maximo_pm25 = lecturas_validas[0]
minimo_pm25 = lecturas_validas[0]

for valor in lecturas_validas:
    if valor > maximo_pm25:
        maximo_pm25 = valor
    if valor < minimo_pm25:
        minimo_pm25 = valor

rango = maximo_pm25 - minimo_pm25

suma_desviaciones = 0.0

for valor in lecturas_validas:
    suma_desviaciones += (valor - promedio_pm25) ** 2

desviacion_estandar = (suma_desviaciones / len(lecturas_validas)) ** 0.5

print("Maximo PM2.5:", maximo_pm25)
print("Minimo PM2.5:", minimo_pm25)
print("Rango PM2.5:", rango)
print("Desviacion estandar:", desviacion_estandar)
```

## Ejercicio 8 — Decaimiento tras cerrar la fuente

Después del pico (58.3), la concentración baja 15% cada hora (se multiplica
por 0.85). Se quiere saber cuántas horas pasan hasta llegar a 12.0 o menos.

Como no se sabe cuántas horas son, se usa un `while`. Dentro del while se
actualiza la concentración para que el bucle termine.

```python
concentracion = 58.3
factor_decaimiento = 0.85
umbral_buena = 12.0

horas_transcurridas = 0

while concentracion > umbral_buena:
    concentracion = concentracion * factor_decaimiento
    horas_transcurridas += 1

print("Horas transcurridas:", horas_transcurridas)
print("Concentracion final:", concentracion)
```

## Ejercicio 9 — Registros de varias estaciones

Hay tres estaciones, cada una en un diccionario con código, sector,
coordenadas (tupla) y lecturas. Solo EST-03 tiene humedad.

Se recorre la lista con un `for`. En cada estación se saca la latitud y la
longitud, se calcula el promedio con otro `for`, y se saca la humedad con
`get` (para que las que no la tienen no den error).

Al final se clasifica el promedio y se imprime una línea por estación.

```python
registros_estaciones = [
    {"codigo": "EST-01", "sector": "Norte",  "coordenadas": (6.28, -75.57), "lecturas": [8.4, 12.7, 15.2]},
    {"codigo": "EST-02", "sector": "Centro", "coordenadas": (6.24, -75.58), "lecturas": [41.8, 58.3, 36.1]},
    {"codigo": "EST-03", "sector": "Sur",    "coordenadas": (6.19, -75.59), "lecturas": [9.6, 13.9, 7.5], "humedad_relativa": 68.0},
]

for estacion in registros_estaciones:
    latitud_est = estacion["coordenadas"][0]
    longitud_est = estacion["coordenadas"][1]

    suma_estacion = 0.0
    for valor in estacion["lecturas"]:
        suma_estacion += valor
    promedio_estacion = suma_estacion / len(estacion["lecturas"])

    humedad = estacion.get("humedad_relativa", "no disponible")

    if promedio_estacion <= 12.0:
        categoria_estacion = "Buena"
    elif promedio_estacion <= 35.4:
        categoria_estacion = "Moderada"
    elif promedio_estacion <= 55.4:
        categoria_estacion = "Dañina para grupos sensibles"
    else:
        categoria_estacion = "Dañina para la salud"

    print(estacion["codigo"], "|", estacion["sector"], "|", latitud_est, "|", longitud_est,
          "|", promedio_estacion, "|", categoria_estacion, "|", humedad)
```
