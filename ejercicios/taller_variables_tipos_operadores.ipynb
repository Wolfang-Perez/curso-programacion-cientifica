### Orden de evaluación (Precedencia)

Python evalúa los operadores lógicos de mayor a menor prioridad según las reglas de **precedencia**:

1. **not** (Máxima prioridad): Se evalúa primero. Convierte not False en True.

> Expresión restante:
> True or True and False

2. **and** (Prioridad media): Se evalúa segundo. Calcula True and False, lo cual resulta en False.

>Expresión restante:
>True or False

3. **or** (Menor prioridad): Se evalúa al final. Calcula True or False, lo cual resulta en el valor exacto:

>True.
