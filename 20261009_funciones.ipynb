```python
def maximo_valor(datos2):
  maximo_valor = -1000
  for dato in datos2:
    if dato >= maximo_valor:
      maximo_valor = dato
  print(f"El máximo valor es {maximo_valor}")
```

```python
import random

datos = [random.randint(0,10000) for _ in range(10)]
print(datos)
maximo_valor(datos)
```
